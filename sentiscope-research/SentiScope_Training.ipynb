# SentiScope NLP training
English sentiment (TweetEval), Xitsonga (AfriSenti), multi-label emotion (GoEmotions). The training script downloads the source data, removes cross-split duplicate texts, compares models using validation macro-F1, and exports compact portable weights. Report remains offline, not in the application UI.


```python
from pathlib import Path
import json, pandas as pd
from IPython.display import Code, display
root = Path.cwd()
if not (root/'train.py').exists(): root = root/'sentiscope-research'
display(Code((root/'train.py').read_text(), language='python'))
```

```python
# Optional: reproduce training after installing dependencies
# import subprocess, sys
# subprocess.run([sys.executable, str(root/'train.py')], check=True)
report = json.loads((root/'report.json').read_text())
for language in ['english','xitsonga_mozambique']:
    print(language, report[language]['counts'])
    display(pd.Series(report[language]['validation_macro_f1']).sort_values(ascending=False))
    display(pd.DataFrame(report[language]['test']).T)
print(report['goemotions'])
print(report['limitations'])
```
